## Άσκηση - Κατανόηση Ενέργειας και Ισχύος

Υπολογίστε στην ```Python``` την ενέργεια ή την ισχύ (ή τίποτε από τα δυο) των παρακάτω σημάτων. Χρησιμοποιήστε τα hints που αναφέρει το Κεφάλαιο για να γλιτώσετε πράξεις:

**(α')** $x[n] = \Big(\frac{1}{2}\Big)^n u[n]$

**(β')** $x[n] = 2\cos\Big(\frac{3\pi}{4}n\Big)$

**(γ')** $x[n] = 3^n u[-n]$

**(δ')** $x[n] = \left\{\begin{array}{lll}
-1, & n < 0 \\
0, & n = 0 \\
1, & n > 0
\end{array}\right.$

Ενδεικτικά, σας δίνεται ο κώδικας για τον υπολογισμό της ισχύος της βηματικής συνάρτησης $u[n]$:

````python
# Set a large number for N
N=5000

# Define signal
n=np.arange(0, N)
x=np.ones(np.size(n))

# Calculating Power P
P = 1/(2*len(x))*np.sum(np.power(x,2))

print('P =' , P)
````


Ο κώδικάς απαντά ότι

    P = 0.5

που είναι και η σωστή απάντηση.

````python
# Insert code here
N = 1000
n=np.arange(0,1001)
x41 = -1*np.heaviside(n, 1)
x42 = 1*np.heaviside(n, 0)
x4 = np.concatenate((x41, x42))
P4 = (1/(2*N+1))*np.sum(np.power(x4,2))
print('P =', P4)

# End of code
````
