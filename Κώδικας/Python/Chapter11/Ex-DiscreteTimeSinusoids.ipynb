````python
import numpy as np
import matplotlib.pyplot as plt
````

## Άσκηση - Ημίτονα και Περιοδικότητα στο διακριτό χρόνο

Έστω το σήμα διακριτού χρόνου 

$$x[n] = 2\cos(0.1\pi n) + 3\sin(0.2\pi n)\tag{1}$$

**(Α')** Ας δημιουργήσουμε σε ```Python``` $50$ δείγματα του παραπάνω διακριτού σήματος και ας εμφανίσουμε το αποτέλεσμα με την εντολή $\texttt{stem}$, ορίζοντας κατάλληλα το διάνυσμα του διακριτού χρόνου $n$, να είναι από μηδέν έως $49$ (σύνολο $50$ χρονικές στιγμές).

````python
n = np.arange(0, 50);                               # Discrete time indices
x = 2*np.cos(0.1*np.pi*n) + 3*np.sin(0.2*np.pi*n);  # Signal

fig = plt.figure(figsize=(12,4))                    # Plot
plt.stem(n, x)                                      # Visualize
plt.title('Synthetic signal')
plt.xlabel('Samples')
plt.show()
````

**i.** Από το σχήμα, μπορείτε να βρείτε ποιά είναι περίπου η περίοδος του σήματος;

### Απάντηση:

**ii.** Υπολογίστε με ακρίβεια την περίοδο του σήματος σε αριθμό δειγμάτων στο χαρτί, και επιβεβαιώστε το αποτέλεσμα ελέγχοντας τις τιμές του σήματος χρησιμοποιώντας ```Python```. Υπολογίσατε σωστά την περίοδο αρχικά "με το μάτι"; 

### Απάντηση:

Έστω τα σήματα $x_{1}[n]=2\cos(0.1\pi n)$ και $x_{2}[n] = 3\sin(0.2\pi n)$. Είναι:
* Για το $x_1: \omega_1 = 0.1\pi, \;\; N_1 = \frac{2\pi}{\omega_1}k = \frac{2\pi}{0.1\pi}k = \frac{2\pi}{\frac{\pi}{10}}k = 20k \xrightarrow{k=1} N_1 = 20$, άρα το σήμα είναι περιοδικό με περίοδο 20 δείγματα.
* Για το $x_2: \omega_2=0.2\pi, N_2 = \frac{2\pi}{\omega_2}k = \frac{2\pi}{0.2\pi}k = 10k \xrightarrow{k=1} N_2 = 10$, άρα το σήμα είναι περιοδικό με περίοδο 10 δείγματα.
$$Ν = \frac{N_1N_2}{Μ.Κ.Δ.\{N_1,N2\}} = \frac{20*10}{Μ.Κ.Δ.\{20,10\}} = \frac{200}{10}\Rightarrow N = 20$$


Άρα, έιχαμε υπολογίσει σωστά την περίοδο στο προηγούμενο ερώτημα “με το μάτι”. Βλέποντας και τις
τιμές που είναι αποθηκευμένες στο σήμα x παρατηρούμε ότι επαναλαμβάνεται το ίδιο μοτίβο
τιμών κάθε 20 δείγματα.

**iii.** Αν αυτό το σήμα έχει προέλθει από δειγματοληψία ενός σήματος συνεχούς χρόνου με συχνότητα δειγματοληψίας $200$ Hz, τότε πόση είναι η περίοδος του σήματος συνεχούς χρόνου; 

### Απάντηση:


Αν αυτό το σήμα έχει προέλθει από δειγματοληψία ενός σήματος συνεχούς χρόνου με συχνότητα δειγματοληψίας $f_s = 200 \; Hz$, τότε η περίοδος του σήματος συνεχούς χρόνου είναι: 

$$\omega_0 = \frac{2\pi f_0}{f_s} \Rightarrow \frac{\pi}{10} = \frac{2\pi f_0}{200} \Rightarrow f_0 = \frac{200\pi}{20\pi}\Rightarrow f_0 = 10Hz \Rightarrow T_0= \frac{1}{f_0} = \frac{1}{10}\Rightarrow \\ T_0 = 0.1 \; sec$$

**(B')** Ας υποθέσουμε ότι προσθέτουμε άλλη μια συνιστώσα στο παραπάνω σήμα:
\begin{equation}
x[n] = 2\cos(0.1\pi n) + 2\cos(0.15\pi n) + 3\sin(0.2\pi n) \tag{2}

**i.** Δημιουργήστε $150$ δείγματα στην ```Python```. Εκτελώντας ξανά την εντολή $\texttt{stem}$ για το παραπάνω σήμα, εμφανίστε το.

````python
# Insert code here
n = np.arange(0, 150);                                                      # Discrete time indices
x = 2*np.cos(0.1*np.pi*n) + 2*np.cos(0.15*np.pi*n) + 3*np.sin(0.2*np.pi*n); # Signal

fig = plt.figure(figsize=(12,4))                                            # Plot
plt.stem(n, x)                                                              # Visualize
plt.title('Synthetic signal')
plt.xlabel('Samples')
plt.show()
# End of code
````

**ii.** Πόση είναι η περίοδος σε αριθμό δειγμάτων;

### Απάντηση:

Έστω τα σήματα $x_1[n] = 2\cos(0.1\pi n), \; x_2[n] = 2\cos(0.15\pi n) \; \text{και} \; x_3[n] = 3sin(0.2\pi n).$ Είναι: 

* Για το $x_1: \omega_1 = 0.1\pi,\; N_1 = \frac{2\pi}{\omega_1}k = \frac{2\pi}{0.1\pi}k = \frac{2\pi}{\frac{\pi}{10}}k \xrightarrow{k=1} N_1 = 20$, άρα το σήμα είναι περιοδικό με περίοδο 20 δείγματα.

* Για το $x_2:\omega_2 = 0.15\pi, \; N_2 = \frac{2\pi}{\omega_2}k = \frac{2\pi}{0.15\pi}k = \frac{2\pi}{\frac{3\pi}{20}}k = \frac{40k}{3} \xrightarrow{k=3} N_2 = 40$, άρα το σήμα είναι περιοδικό με περίοδο 40 δείγματα.

* Για το $x_3:\omega_3=0.2\pi, \; N_3 = \frac{2\pi}{\omega_3}k = \frac{2\pi}{0.2\pi}k =  \frac{2\pi}{\frac{\pi}{5}}k = 10k \xrightarrow{k=1} N_3 = 10$, άρα το σήμα είναι περιοδικό με περίοδο 10 δείγματα.

Άρα, γνωρίζουμε βάσει της θεωρίας ότι το σήμα $x[n] = x_1[n] + x_2[n] + x_3[n]$ θα είναι επίσης περιοδικό με περίοδο 

$$N_{12} = \frac{N_1N_2}{Μ.Κ.Δ.\{N_1,N2\}} = \frac{20\cdot 40}{20} = 40$$

και

$$N = \frac{N_{12}N_3}{Μ.Κ.Δ\{N_{12}, N_3\}} = \frac{40 \cdot 10}{10} = 40$$

**iii.** Πόση είναι η περίοδος σε δευτερόλεπτα, αν το σήμα συνεχούς χρόνου από το οποίο προήλθε δειγματοληπτήθηκε με συχνότητα δειγματοληψίας $200 \; Hz$;

### Απάντηση:

Αν αυτό το σήμα έχει προέλθει από δειγματοληψία ενός σήματος συνεχούς χρόνου με συχνότητα δειγματοληψίας $f_s = 200 \; Hz$, τότε η περίοδος του σήματος συνεχούς χρόνου είναι:

$$\omega_0 = \frac{2\pi}{N} = \frac{2\pi f_0}{f_s}\Rightarrow \frac{\pi}{20} = \frac{2\pi f_0}{200} \Rightarrow f_0 = \frac{200\pi}{40\pi} \Rightarrow f_0= 5 \; Hz$$

και 

\begin{equation}
T_0 = \frac{1}{f_0} \Rightarrow T_0 = \frac{1}{5} \text{ s}
\end{equation}
