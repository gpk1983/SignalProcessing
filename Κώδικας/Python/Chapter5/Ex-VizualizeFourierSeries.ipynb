```python
import numpy as np                        # Μαθηματική/αριθμητική βιβλιοθήκη
import matplotlib.pyplot as plt           # Βιβλιοθήκη απεικονίσεων
import ipywidgets as widgets              # Διαδραστικά widgets
from ipywidgets import interact, IntSlider
%matplotlib inline
```

## Άσκηση - Οπτικοποίηση Σειρών Fourier

Αν λύσατε σωστά την άσκηση $\texttt{Ex-FourierSeries}$, θα είδατε πόσο καλά δουλεύει η προσέγγιση μιας σειράς Fourier με μόλις 60 όρους. Όμως είδαμε μόνο το τελικό αποτέλεσμα!

Δε θα ήταν πολύ όμορφο να δούμε σιγά σιγά, όρο προς όρο, το πως δημιουργείται αυτό το αποτέλεσμα? 😊😊

Απλά τρέξτε τον παρακάτω κώδικα και "παίξτε" με τον slider στο αποτέλεσμά του.

```python
def fourier_series(t, A, f0, n_terms):
    """Υπολογίζουμε την σειρά Fourier ενός πριονωτού σήματος για n_terms όρους."""
    # Ξεκινάμε από τη μέση τιμή X0 = A/2 (ο όρος k=0)
    y = np.full_like(t, A / 2)
    # Προσθέτουμε έναν–έναν τους n_terms αρμονικούς όρους
    for k in range(1, n_terms + 1):
        y += (A / (np.pi * k)) * np.cos(2 * np.pi * k * f0 * t + np.pi / 2)
    return y

# Επιλέγουμε μια διαμέριση του άξονα του χρόνου
dt = 0.001
# Ας επιλέξουμε την περίοδο, και άρα τη θεμελιώδη συχνότητα, του περιοδικού σήματος
T0 = 1
f0 = 1 / T0

# Η παράμετρος Α
A = 1

# Κατασκευάζουμε τον άξονα του χρόνου: 4 περίοδοι, στο [-2Τ0, 2Τ0]
t = np.arange(start=-2 * T0, stop=2 * T0, step=dt)


def plot_fourier(n_terms=1):
    """Σχεδιάζουμε τη σειρά Fourier για n_terms όρους."""
    y = fourier_series(t, A, f0, n_terms)

    fig, ax = plt.subplots(figsize=(12, 4))
    ax.plot(t, y, label=f"{n_terms} όροι")
    ax.set_title("Σειρά Fourier")
    ax.set_xlabel("Χρόνος")
    ax.set_ylabel("Πλάτος")
    # Λίγο περιθώριο ώστε να φαίνεται το overshoot (φαινόμενο Gibbs)
    ax.set_ylim(-0.1, 1.1)
    ax.legend(loc="upper right")
    plt.show()


# Ο slider καλεί την plot_fourier κάθε φορά που αλλάζει η τιμή του.
interact(
    plot_fourier,
    n_terms=IntSlider(min=1, max=60, step=1, value=1, description="Όροι"),
);
```

Γράψτε παρόμοιο κώδικα για το περιοδικό σήμα που αναπτύσσεται σε σειρά Fourier ως

$ x(t) = \displaystyle \sum_{k=1}^{+\infty} \frac{1}{2\pi k^2} \cos(2\pi k f_0 t) $

```python
def fourier_series_2(t, f0, n_terms):
    """Υπολογίζουμε την σειρά Fourier ενός πριονωτού σήματος για n_terms όρους."""
    # Ξεκινάμε από τη μέση τιμή X0 = 0 (ο όρος k=0)
    y = np.full_like(t, 0)
    # Προσθέτουμε έναν–έναν τους n_terms αρμονικούς όρους
    for k in range(1, n_terms + 1):
        y += (1 / (2*np.pi*k**2)) * np.cos(2 * np.pi * k * f0 * t)
    return y

# Επιλέγουμε μια διαμέριση του άξονα του χρόνου
dt = 0.001
# Ας επιλέξουμε την περίοδο, και άρα τη θεμελιώδη συχνότητα, του περιοδικού σήματος
T0 = 1
f0 = 1 / T0

# Κατασκευάζουμε τον άξονα του χρόνου: 4 περίοδοι, στο [-2Τ0, 2Τ0]
t = np.arange(start=-2 * T0, stop=2 * T0, step=dt)


def plot_fourier_2(n_terms=1):
    """Σχεδιάζουμε τη σειρά Fourier για n_terms όρους."""
    y = fourier_series_2(t, f0, n_terms)

    fig, ax = plt.subplots(figsize=(12, 4))
    ax.plot(t, y, label=f"{n_terms} όροι")
    ax.set_title("Σειρά Fourier")
    ax.set_xlabel("Χρόνος")
    ax.set_ylabel("Πλάτος")
    # Λίγο περιθώριο ώστε να φαίνεται το overshoot (φαινόμενο Gibbs)
    #ax.set_ylim(-0.1, 1.1)
    ax.legend(loc="upper right")
    plt.show()


# Ο slider καλεί την plot_fourier κάθε φορά που αλλάζει η τιμή του.
interact(
    plot_fourier_2,
    n_terms=IntSlider(min=1, max=60, step=1, value=1, description="Όροι"),
);
```
