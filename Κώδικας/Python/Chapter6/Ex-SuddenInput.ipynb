```python
import numpy as np                        # Μαθηματική/αριθμητική βιβλιοθήκη
import matplotlib.pyplot as plt           # Βιβλιοθήκη απεικονίσεων
%matplotlib inline
```

## Άσκηση - Το φαινόμενο της ξαφνικής εισόδου (*sudden input*) σε ΓΧΑ σύστημα

Όταν αναλύουμε ένα Γραμμικό Χρονικά Αμετάβλητο (ΓΧΑ) σύστημα στο πεδίο της συχνότητας, χρησιμοποιούμε την απόκριση συχνότητας $H(f)$ και λέμε ότι 

**"αν βάλω στην είσοδο ένα ημίτονο συχνότητας $f_0$, στην έξοδο θα πάρω το ίδιο ημίτονο, πολλαπλασιασμένο κατά $|H(f_0)|$ και μετατοπισμένο κατά $\angle H(f_0)$".** 

Αυτό όμως κρύβει μια σιωπηρή παραδοχή: ότι το ημίτονο υπάρχει **από πάντα**, δηλ. από $t=-\infty$! 😅

Στην πράξη όμως εμείς **ανοίγουμε τον διακόπτη** κάποια στιγμή - ας πούμε στο $t=0$. Εφαρμόζουμε δηλαδή την είσοδο *απότομα/ξαφνικά*. Αυτή η ξαφνική εφαρμογή δεν διεγείρει μόνο τη **σταθερής κατάστασης απόκριση** (steady-state) που προβλέπει η $H(f_0)$, αλλά και την περίφημη **μεταβατική απόκριση** του συστήματος, η οποία - για ένα ευσταθές σύστημα - σβήνει με την πάροδο του χρόνου. 

Το αποτέλεσμα είναι ότι η έξοδος **δεν** ξεκινά κατ' ευθείαν ως το "καθαρό" ημίτονο της μόνιμης κατάστασης - χρειάζεται λίγο χρόνο μέχρι να "κάτσει". Αυτό ακριβώς είναι το **φαινόμενο της ξανφικής εισόδου**, και θα το δούμε με τα μάτια μας σε αυτήν την άσκηση.

Ας δουλέψουμε με το **ίδιο** σύστημα που είδαμε στην άσκηση των διαφορικών εξισώσεων μερικά κεφάλαια νωρίτερα (ένα κύκλωμα $RC$), το οποίο περιγράφεται από τη διαφορική εξίσωση

$$ \frac{d}{dt}y(t) + 4y(t) = 4x(t), \qquad y(0^-)=0 $$

Με μηδενικές αρχικές συνθήκες, το σύστημα είναι ΓΧΑ. Παίρνοντας μετασχ. Fourier βρίσκουμε τη συχνοτική απόκριση:

$$H(f) = \frac{4}{4+j2\pi f} $$

Το σύστημα έχει μια μεταβατική απόκριση της μορφής $e^{-4t}$ - δεν ξέρετε ακόμα γιατί, αλλά κρατήστε το αυτό στο μυαλό σας, θα το ξαναδούμε! 😉

**Είσοδος που εφαρμόζεται απότομα:** Έστω ότι στο $t=0$ ανοίγουμε τον διακόπτη και εφαρμόζουμε ένα συνημίτονο:

$$ x(t) = \cos(\omega_0 t)\, u(t) $$

Το $u(t)$ (βηματική συνάρτηση) είναι ακριβώς αυτό που κάνει την είσοδο *απότομη*: πριν το $t=0$ δεν υπάρχει τίποτα, και μετά εμφανίζεται ξαφνικά το ημίτονο.

Αν το ημίτονο υπήρχε από το $-\infty$, η έξοδος θα ήταν **μόνο** η μόνιμη απόκριση, την οποία μας δίνει κατ' ευθείαν η απόκριση συχνότητας:

$$ y_{ss}(t) = |H(f_0)| \, \cos\!\left(2\pi f_0 t + \angle H(f_0)\right) $$

όπου, για το δικό μας $H(f)=\dfrac{4}{4+j2\pi f}$,

$$ |H(f_0)| = \frac{4}{\sqrt{16+(2\pi f_0)^2}}, \qquad \angle H(f_0) = -\arctan\!\left(\frac{\pi f_0}{2}\right) $$

**Η συνολική απόκριση:** Επειδή όμως η είσοδος εφαρμόζεται "ξαφνικά", η πλήρης λύση της διαφορικής εξίσωσης είναι το άθροισμα της μόνιμης απόκρισης **και** μιας μεταβατικής συνιστώσας της μορφής $C e^{-4t}$:

$$ y(t) = \underbrace{|H(f_0)|\cos\!\left(2\pi f_0 t + \angle H(f_0)\right)}_{\text{απόκριση σταθερής κατάστασης } y_{ss}(t)} \; + \; \underbrace{C\,e^{-4t}}_{\text{μεταβατική απόκριση}}, \qquad t \ge 0 $$

Τη σταθερά $C$ τη βρίσκουμε από την αρχική συνθήκη $y(0)=0$ (το σύστημα ξεκινά ηρεμία):

$$ y(0) = 0 \;\Longrightarrow\; |H(f_0)|\cos\left(\angle H(f_0)\right) + C = 0 \;\Longrightarrow\; C = -\,|H(f_0)|\cos\left(\angle H(f_0)\right) = -\,\operatorname{Re}\{H(f_0)\} $$

και για το δικό μας σύστημα $\operatorname{Re}\{H(f_0)\} = \dfrac{16}{16+(2\pi f_0)^2}$, άρα

\begin{equation}
y(t) = |H(f_0)|\cos\!\left(2\pi f_0 t + \angle H(f_0)\right) \; - \; \frac{16}{16+4\pi^2 f_0^2}\, e^{-4t}, \qquad t\ge 0
\end{equation}

Η δεύτερη - εκθετικά φθίνουσα - συνιστώσα είναι ακριβώς το "τίμημα" της ξαφνικής εισόδου. Μετά από λίγους χρόνους $\tau = 1/4$ s θα έχει σβήσει και η έξοδος θα ταυτιστεί με την απόκριση σταθερής κατάστασης.

Ας το επιβεβαιώσουμε αριθμητικά. Θα προσομοιώσουμε το "αληθινό" σύστημα διακριτοποιώντας τη διαφορική εξίσωση με τη **μέθοδο του Euler** (ακριβώς όπως σε προηγούμενη άσκηση προηγούμενου Κεφαλαίου - των διαφορικών εξισώσεων):

$$ y(n+1) = y(n) - 4T_s\,y(n) + 4T_s\,x(n) $$

Η προσομοίωση αυτή περιέχει **αυτόματα** και τη μεταβατική και τη μόνιμη συνιστώσα (δεν "ξέρει" τίποτα για τη διάσπαση που κάναμε "στα χαρτιά"). Θα τη συγκρίνουμε με (α) τη θεωρητική ολική λύση - εξίσωση $(1)$ - και (β) **μόνο** με τη μόνιμη απόκριση $y_{ss}(t)$, ώστε να δούμε πόσο διαφέρουν στην αρχή.

```python
f0 = 3                                    # Συχνότητα εισόδου (Hz)
w0 = 2*np.pi*f0                           # Κυκλική συχνότητα (rad/s)

Ts = 0.001                               # Περίοδος δειγματοληψίας (μικρή, για καλή προσέγγιση Euler)
t = np.arange(start=0, stop=3, step=Ts)  # Διάνυσμα χρόνου t

x = np.cos(w0*t)                         # Ξαφνική είσοδος x(t) = cos(2πf0 t) u(t) (το u(t) υπονοείται: t ξεκινά από 0)

# Αριθμητική ("αληθινή") έξοδος μέσω Euler: περιέχει μεταβατική + μόνιμη
y = np.zeros(len(t))                      # Δέσμευση μνήμης
y[0] = 0                                  # Αρχική συνθήκη (ηρεμία)
for n in range(0, len(y)-1):              # Υλοποίηση διακριτοποιημένης διαφορικής εξίσωσης
    y[n+1] = y[n] - 4*Ts*y[n] + 4*Ts*x[n]

# Απόκριση συχνότητας στο w0 
H = 4.0 / (4.0 + 1j*w0)                   # H(f0)
Hmag = np.abs(H)                          # |H(f0)|
Hphase = np.angle(H)                      # <H(f0)

# Μόνιμη απόκριση (steady-state), σαν να υπήρχε το ημίτονο από πάντα
y_ss = Hmag*np.cos(w0*t + Hphase)

# Θεωρητική ΟΛΙΚΗ λύση: μόνιμη + μεταβατική (εξίσωση (1) στην εκφώνηση)
C = -np.real(H)                          # C = -Re{H(f0)} = -16/(16+4πf0^2)
y_theory = y_ss + C*np.exp(-4*t)

# Η μεταβατική συνιστώσα μόνη της
y_trans = C*np.exp(-4*t)
```

Ας τα απεικονίσουμε όλα μαζί.

```python
plt.figure(figsize=(12,8))

# Πάνω: αριθμητική έξοδος vs μόνιμη απόκριση
plt.subplot(2,1,1)
plt.plot(t, y, '-', label='Αριθμητική έξοδος (Euler) - ΟΛΙΚΗ')
plt.plot(t, y_ss, '--', label='Απόκριση σταθερής κατάστασης $y_{ss}(t)$ μόνο')
plt.plot(t, y_theory, ':', color='k', label='Θεωρητική ολική λύση - εξ. (1)')
plt.grid(linestyle=':')
plt.title('Φαινόμενο ξαφνικής εισόδου: η έξοδος δεν ξεκινά ως καθαρό ημίτονο')
plt.ylabel('Πλάτος')
plt.legend(loc='upper right')

# Κάτω: η μεταβατική συνιστώσα (διαφορά ολικής - μόνιμης)
plt.subplot(2,1,2)
plt.plot(t, y - y_ss, '-', label='Ολική $-$ σταθερής κατάστασης (αριθμητικά)')
plt.plot(t, y_trans, '--', color='tab:red', label='$C e^{-4t}$ (θεωρία)')
plt.grid(linestyle=':')
plt.title('Η μεταβατική συνιστώσα σβήνει εκθετικά')
plt.xlabel('Χρόνος (s)')
plt.ylabel('Πλάτος')
plt.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

Παρατηρήστε δύο πράγματα:

- Στην αρχή ($t$ κοντά στο $0$) η αληθινή έξοδος **αποκλίνει** από την απόκριση σταθερής κατάστασης $y_{ss}(t)$. Αυτή η απόκλιση είναι η μεταβατική συνιστώσα - το αποτύπωμα του ότι "ανοίξαμε τον διακόπτη" απότομα.
- Στο κάτω γράφημα, η διαφορά *ολική $-$ σταθερής κατάστασης* ταυτίζεται με την εκθετική $C e^{-4t}$ και σβήνει μέσα σε λίγα $\tau = 1/4$ s. Μετά από αυτό, η έξοδος είναι πρακτικά ίδια με το "καθαρό" ημίτονο της σταθερής κατάστασης.

---

Το πόσο "εμφανές" είναι το φαινόμενο εξαρτάται από τη σχέση της συχνότητας εισόδου $f_0$ με την τιμή στο $-4$. Γράψτε κώδικα `Python` που:

(α') Επαναλαμβάνει την προσομοίωση για μια **μεγαλύτερη** συχνότητα εισόδου, π.χ. $f_0 = 15$ Hz, και τυπώνει τα αντίστοιχα γραφήματα (ολική έξοδος, απόκριση σταθερής κατάστασης, και μεταβατική).

(β') Στο ίδιο γράφημα, σχεδιάστε και την **περιβάλλουσα** $\pm|C|e^{-4t}$ της μεταβατικής απόκρισης, για να επιβεβαιώσετε ότι η μεταβατική απόκριση "χωράει" μέσα σε αυτήν.

*Υπόδειξη:* αρκεί να αλλάξετε το `f0` και να επαναχρησιμοποιήσετε τον κώδικα παραπάνω.

```python
# INSERT CODE HERE
f0 = 15                                  # Μεγαλύτερη συχνότητα εισόδου (Hz)
w0 = 2*np.pi*f0

Ts = 0.0005                              # Ακόμη μικρότερο Ts για καλή προσέγγιση σε ψηλή συχνότητα
t = np.arange(start=0, stop=1.5, step=Ts)

x = np.cos(w0*t)                         # Απότομη είσοδος

# Αριθμητική έξοδος (Euler)
y = np.zeros(len(t))
y[0] = 0
for n in range(0, len(y)-1):
    y[n+1] = y[n] - 4*Ts*y[n] + 4*Ts*x[n]

# Απόκριση συχνότητας, μόνιμη και μεταβατική
H = 4.0 / (4.0 + 1j*w0)
y_ss = np.abs(H)*np.cos(w0*t + np.angle(H))
C = -np.real(H)
y_trans = C*np.exp(-4*t)

# Γραφήματα
plt.figure(figsize=(12,8))

plt.subplot(2,1,1)
plt.plot(t, y, '-', label='Αριθμητική έξοδος (Euler) - ΟΛΙΚΗ')
plt.plot(t, y_ss, '--', label='Απόκριση σταθερής κατάστασης $y_{ss}(t)$ μόνο')
plt.grid(linestyle=':')
plt.title(f'Ξαφνική είσοδος για f0 = {f0} Hz')
plt.ylabel('Πλάτος')
plt.legend(loc='upper right')

plt.subplot(2,1,2)
plt.plot(t, y - y_ss, '-', label='Ολική $-$ σταθερής κατάστασης (αριθμητικά)')
plt.plot(t, y_trans, '--', color='tab:red', label='$C e^{-4t}$ (θεωρία)')
plt.plot(t,  np.abs(C)*np.exp(-4*t), ':', color='k', label='Περιβάλλουσα $\\pm|C|e^{-4t}$')
plt.plot(t, -np.abs(C)*np.exp(-4*t), ':', color='k')
plt.grid(linestyle=':')
plt.title('Μεταβατική συνιστώσα και η περιβάλλουσά της')
plt.xlabel('Χρόνος (s)')
plt.ylabel('Πλάτος')
plt.legend(loc='upper right')

plt.tight_layout()
plt.show()

# END OF CODE
```

Τι παρατηρείτε καθώς μεγαλώνει η συχνότητα εισόδου $f_0$; Πώς αλλάζει το πλάτος της μόνιμης απόκρισης και πόσο "εμφανής" γίνεται η μεταβατική; Γράψτε την απάντησή σας σε αυτό το κελί:

### Απάντηση:

Καθώς μεγαλώνει η $f_0$, το $|H(f_0)| = 4/\sqrt{16+4\pi^2 f_0^2}$ **μικραίνει**: το σύστημα (ένα χαμηλοπερατό $RC$) εξασθενεί όλο και περισσότερο τις υψηλές συχνότητες, οπότε η μόνιμη απόκριση έχει πολύ μικρό πλάτος. Ταυτόχρονα η σταθερά $C=-16/(16+4\pi^2 f_0^2)$ επίσης μικραίνει, δηλ. η μεταβατική συνιστώσα γίνεται μικρότερη σε πλάτος - **αλλά** ο ρυθμός απόσβεσής της ($e^{-4t}$) παραμένει ίδιος, γιατί εξαρτάται μόνο από το σύστημα και όχι από τη συχνότητα εισόδου. Έτσι, σε υψηλές συχνότητες η μεταβατική απόκριση μπορεί να είναι **συγκρίσιμη ή και μεγαλύτερη** από τη μικροσκοπική πλέον απόκριση σταθερής κατάστασης, κάνοντας το ξεκίνημα της εξόδου να φαίνεται "παραμορφωμένο" για περισσότερους (σχετικά) κύκλους πριν "κάτσει". Σε κάθε περίπτωση, μετά από λίγα $\tau = 1/4$ s η μεταβατική απόκριση σβήνει και μένει μόνο η απόκριση σταθερής κατάστασης.
