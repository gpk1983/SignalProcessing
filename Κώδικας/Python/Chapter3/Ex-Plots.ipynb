```python
import numpy as np                 # εισάγουμε τη βιβλιοθήκη numpy και θα τη "φωνάζουμε" np
import matplotlib.pyplot as plt    # εισάγουμε τη συνάρτηση pyplot της βιβλιοθήκης matplotlib και θα τη "φωνάζουμε" plt
```

## Άσκηση - Γραφήματα πραγματικών συναρτήσεων

Η απεικόνιση συναρτήσεων στην `Python` είναι εύκολη. Αν για παράδειγμα θέλετε να απεικονίσετε ένα ημίτονο της μορφής

$$ x(t) = 4\cos(2\pi t + \pi/3)$$
    
στο διάστημα $[-1,1)$, ο παρακάτω κώδικας θα το κάνει για σας. 

Μελετήστε τον γραμμή-γραμμή και τρέξτε τον (εκτελέστε το παρακάτω κελί επιλέγοντάς το και πατώντας Shift + Enter).

```python
t = np.arange(start=-1, stop=1, step=0.01)   # Ορίζουμε τον άξονα του χρόνου: ξεκινάμε από το -1, σταματάμε στο 1, με βήμα 0.01. Προσέξτε, το σημείο 1 δεν περιλαμβάνεται, δηλ.
                                             # το διάστημα είναι [-1,1), δηλ. το [-1, 1-step]
x = 4 * np.cos(2*np.pi*t + np.pi/3)          # Ορίζουμε τη συνάρτησή μας

plt.figure(figsize=(12,3))            # Ορίζουμε ότι θέλουμε να φτιάξουμε ένα γράφημα
plt.plot(t, x, '-', color='tab:blue') # Φτιάχνουμε το γράφημα
plt.xlim([min(t), max(t)])            # Ομορφαίνουμε
plt.grid(linestyle=':')               # --||--
plt.title('Ένα απλό ημιτονοειδές')    # --||--
plt.xlabel('Χρόνος (s)')              # --||--
plt.show()
```

Μια μικρή επεξήγηση: δε θα μπορούσαμε να αποθηκεύσουμε όλες τις τιμές μιας συνάρτησης συνεχούς χρόνου, καθώς αυτές είναι άπειρες και η χωρητικότητα του υπολογιστή μας πεπερασμένη. Αυτό που κάνουμε στο διάνυσμα $t$ είναι να πάρουμε <em>δείγματα</em> από το συνεχή άξονα $t$. Ξεκινάμε από τον αριθμό $t=-1$ και διανύουμε τον άξονα με βήμα $dt=0.01$, καταλήγοντας στη χρονική στιγμή $t=1$ (χωρίς να συμπεριλαμβάνουμε το σημείο, όταν χρησιμοποιούμε τη συνάρτηση `arange` της `NumPy`). 

Στη συνέχεια ζητάμε από την `Python` να υπολογίσει για μας τις τιμές της ημιτονοειδούς συνάρτησης στις χρονικές στιγμές που υπάρχουν στο διάνυσμα $t$. Η συνάρτηση `plt.plot` της `matplotlib` αναλαμβάνει την απεικόνιση σε γράφημα, ενώ οι υπόλοιπες εντολές είναι διακοσμητικού χαρακτήρα (αλλά πολύ επεξηγηματικές για αυτόν/ήν που διαβάζει το γράφημα).

Τυπώστε στην `Python` τις συναρτήσεις:

(i') $$x(t) = t^3 e^{-3t}, \ t \in [0,5)$$

(ii') $$x(t) = 2\cos(2\pi 20t + \pi/2) + \sin(2\pi 10t - \pi/3), \ t \in [-3,3)$$

(iii') $$ x(t) = 2t^2 + 5t + 6, \ t \in [-5, 5)$$

(iv') $$ x(t) = \frac{e^{t}}{t^4-1}, \ t \in [0,2)$$

Γράψτε σε σχόλιο παρακάτω τι παρατηρείτε ότι συμβαίνει στο γράφημα της τελευταίας περίπτωσης. Χρησιμοποιήστε το παρακάτω κελί για τον κώδικα...

```python
# (i)
t = np.arange(0, 5, 0.01)             # Ορίζουμε τον άξονα του χρόνου: ξεκινάμε από το 0, σταματάμε στο 5, με βήμα 0.01. Προσέξτε, το σημείο 5 δεν περιλαμβάνεται.
x = t**3 * np.exp(-3*t)               # Ορίζουμε τη συνάρτησή μας

plt.figure(figsize=(12,3))            # Ορίζουμε ότι θέλουμε να φτιάξουμε ένα γράφημα
plt.plot(t, x, '-', color='tab:blue') # Φτιάχνουμε το γράφημα
plt.xlim([min(t), max(t)])            # Ομορφαίνουμε
plt.grid(linestyle=':')               # --||--
plt.title('(a)')                      # --||--
plt.xlabel('Time (s)')                # --||--
```

```python
# (ii)
t = np.arange(-3, 3, 0.01)            # Ορίζουμε τον άξονα του χρόνου: ξεκινάμε από το -3, σταματάμε στο 3, με βήμα 0.01. Προσέξτε, το σημείο 3 δεν περιλαμβάνεται.
x = 2*np.cos(2*np.pi*20*t + np.pi/2) + np.sin(2*np.pi*10*t - np.pi/3)                # Ορίζουμε τη συνάρτησή μας

plt.figure(figsize=(12,3))            # Ορίζουμε ότι θέλουμε να φτιάξουμε ένα γράφημα
plt.plot(t, x, '-', color='tab:blue') # Φτιάχνουμε το γράφημα
plt.xlim([min(t), max(t)])            # Ομορφαίνουμε
plt.grid(linestyle=':')               # --||--
plt.title('(b)')                      # --||--
plt.xlabel('Time (s)')                # --||--
```

```python

# (iii)
t = np.arange(-5, 5, 0.01)           # Ορίζουμε τον άξονα του χρόνου: ξεκινάμε από το -5, σταματάμε στο 5, με βήμα 0.01. Προσέξτε, το σημείο 5 δεν περιλαμβάνεται.
x = 2*t**2 + 5*t + 6                 # Ορίζουμε τη συνάρτησή μας

plt.figure(figsize=(12,3))            # Ορίζουμε ότι θέλουμε να φτιάξουμε ένα γράφημα
plt.plot(t, x, '-', color='tab:blue') # Φτιάχνουμε το γράφημα
plt.xlim([min(t), max(t)])            # Ομορφαίνουμε
plt.grid(linestyle=':')               # --||--
plt.title('(c)')                      # --||--
plt.xlabel('Time (s)')                # --||--
```

```python
# (iv)
t = np.arange(0, 2, 0.01)             # Ορίζουμε τον άξονα του χρόνου: ξεκινάμε από το 0, σταματάμε στο 2, με βήμα 0.01. Προσέξτε, το σημείο 2 δεν περιλαμβάνεται.
x = np.exp(t) / (t**4 - 1)            # Ορίζουμε τη συνάρτησή μας

plt.figure(figsize=(12,3))            # Ορίζουμε ότι θέλουμε να φτιάξουμε ένα γράφημα
plt.plot(t, x, '-', color='tab:blue') # Φτιάχνουμε το γράφημα
plt.xlim([min(t), max(t)])            # Ομορφαίνουμε
plt.grid(linestyle=':')               # --||--
plt.title('(d)')                      # --||--
plt.xlabel('Time (s)')                # --||--
```
