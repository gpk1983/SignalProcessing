```python
import numpy as np                 # εισάγουμε τη βιβλιοθήκη numpy και θα τη "φωνάζουμε" np
import matplotlib.pyplot as plt    # εισάγουμε τη συνάρτηση pyplot της βιβλιοθήκης matplotlib και θα τη "φωνάζουμε" plt
```

## Άσκηση - Γραφήματα Μιγαδικών Συναρτήσεων μιας μεταβλητής $t$

Ο παρακάτω κώδικας εκτυπώνει τμήμα του πραγματικού μέρους της γραφικής παράστασης του *μιγαδικού* σήματος

$$x(t)=\mathrm{Re}\{(1+2j)e^{j2\pi 200t}\}$$

παίρνοντας τιμές του σήματος $x(t)$ ανά $1/6000$ δευτερόλεπτα στο διάστημα $[-0.05, 0.15]$ s. Δηλ. $6000$ τιμές της συνάρτησης $x(t)$ σε ένα δευτερόλεπτο!

Διαβάστε το πιο κάτω κελί, επιλέξτε το, και πατήστε Shift + Enter για να το τρέξετε.

```python
f0 = 200                                                   # μια τιμή για τη συχνότητα του εκθετικού
Z = (1+2j)                                                 # ο phasor του μιγαδικού εκθετικού

dt = 1/6000                                                # ο ρυθμός που παίρνουμε δείγματα από τη συνεχή συνάρτηση x(t)
t = np.arange(start=-0.05, stop=0.15, step=dt)             # ο άξονας του χρόνου: από -0.05 ως 0.15, με "βήμα" dt

xt = (Z * np.exp(1j*2*np.pi*f0*t)).real                    # η συνάρτησή μας!

plt.figure(figsize=(12,3))                                 # Απεικόνιση: ανοίγουμε ένα figure
plt.plot(t, xt)                                            # Καλούμε την plt.plot για την απεικόνιση
plt.grid(linestyle=':')                                    # Ομορφαίνουμε 
plt.xlabel('Time (s)')                                     # Ομορφαίνουμε 
plt.title('Real part of signal x(t)')                      # Ομορφαίνουμε  
plt.xlim([min(t), max(t)])                                 # Ομορφαίνουμε 
plt.show()                                                 # Ομορφαίνουμε 
```

Διαβάστε λοιπόν προσεκτικά τον παραπάνω κώδικα και γράψτε κώδικα που τυπώνει στο ίδιο διάστημα και με το ίδιο $dt$:

(α') το φανταστικό μέρος του παραπάνω σήματος
$$x(t)=\mathrm{Im}\{(1+2j)e^{j2\pi 200t}\}$$

(β') το σήμα
$$y(t) = e^{-4t}\cos\left(2\pi 1200t + \frac{\pi}{3} \right)$$

(γ') το πραγματικό και το φανταστικό μέρος του σήματος
$$z(t) = 2e^{-6t}e^{j(2\pi 50t+ \pi/12)}$$

```python
dt = 1/6000                                            # Σας δίνουμε το "βημα" ...
t = np.arange(start=-0.05, stop=0.15, step=dt)         # ...και τον άξονα του χρόνου t

# (a') 
xx = (Z * np.exp(1j*2*np.pi*f0*t)).imag

plt.figure(figsize=(12,3))                                 # Απεικόνιση: ανοίγουμε ένα figure
plt.plot(t, xx)                                            # Καλούμε την plt.plot για την απεικόνιση
plt.grid(linestyle=':')                                    # Ομορφαίνουμε 
plt.xlabel('Time (s)')                                     # Ομορφαίνουμε 
plt.title('Imaginary part of signal x(t)')                 # Ομορφαίνουμε  
plt.xlim([min(t), max(t)])                                 # Ομορφαίνουμε 
plt.show()
```

```python
# (b')

yy = np.exp(-4*t) * np.cos(2*np.pi*1200*t + np.pi/3)

plt.figure(figsize=(12,3))                                 # Απεικόνιση: ανοίγουμε ένα figure
plt.plot(t, yy)                                            # Καλούμε την plt.plot για την απεικόνιση
plt.grid(linestyle=':')                                    # Ομορφαίνουμε 
plt.xlabel('Time (s)')                                     # Ομορφαίνουμε 
plt.title('Signal y(t)')                                   # Ομορφαίνουμε  
plt.xlim([min(t), max(t)])                                 # Ομορφαίνουμε 
plt.show()
```

```python
# (c')

zz = 2*np.exp(-6*t) * np.exp(1j*(2*np.pi*50*t + np.pi/12))

# Τυπώνουμε το πραγματικό μέρος...

plt.figure(figsize=(12,3))                                 # Απεικόνιση: ανοίγουμε ένα figure
plt.plot(t, zz.real)                                       # Καλούμε την plt.plot για την απεικόνιση
plt.grid(linestyle=':')                                    # Ομορφαίνουμε 
plt.xlabel('Time (s)')                                     # Ομορφαίνουμε 
plt.title('Real part of signal z(t)')                      # Ομορφαίνουμε  
plt.xlim([min(t), max(t)])                                 # Ομορφαίνουμε 
plt.show()

# ...και το φανταστικό μέρος

plt.figure(figsize=(12,3))                                 # Απεικόνιση: ανοίγουμε ένα figure
plt.plot(t, zz.imag)                                       # Καλούμε την plt.plot για την απεικόνιση
plt.grid(linestyle=':')                                    # Ομορφαίνουμε 
plt.xlabel('Time (s)')                                     # Ομορφαίνουμε 
plt.title('Imaginary part of signal z(t)')                 # Ομορφαίνουμε  
plt.xlim([min(t), max(t)])                                 # Ομορφαίνουμε 
plt.show()
```
