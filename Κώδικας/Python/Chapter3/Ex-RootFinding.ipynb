## Άσκηση - Υπολογισμός ριζών πολυωνύμων

Η συνάρτηση `roots` της `NumPy` υπολογίζει ρίζες πολυωνύμων. Για παράδειγμα, για το πολυώνυμο
<br>
$$ x^2+3x+2$$
    
η σύνταξη είναι η παρακάτω: (εκτελέστε το παρακάτω κελί επιλέγοντάς το και πατώντας Shift + Enter)

```python
coeffs = [1, 3, 2]              # συντελεστές πολυωνύμου x^2 + 3x + 2
r = np.roots(coeffs)            # κλήση της συνάρτησης roots, της βιβλιοθήκης numpy

print(f"Οι ρίζες του πολυωνύμου είναι οι r = {r}")
```

Επιβεβαιώστε τις απαντήσεις των παρακάτω εξισώσεων με χρήση της `Python`.

- $z^3 + 8-8j = 0$, $\: \: z=(8\sqrt{2})^{1/3}e^{j(2\pi k/3 + \pi/4)}, k =0,1,2$

- $z^4 = 16-16j$, $\: \: z=(16\sqrt{2})^{1/4}e^{j(\pi k/2 - \pi/16)}, k =0,1,2,3$

- $z^5 + 32j = 0$, $\: \: z=(32)^{1/5}e^{j(2\pi k/5 - \pi/10)}, k =0,1,2,3,4$

Θα παρατηρήσετε ότι σε κάποιες από τις απαντήσεις σας θα υπάρχουν διαφορές σε σχέση με τη θεωρητική λύση. Επιβεβαιώστε όμως ότι αυτό συμβαίνει σε αριθμούς ΠΑΡΑ ΠΟΛΥ μικρούς ($\approx 10^{-16}$, που στην ``Python`` εμφανίζονται στη μορφή $e^{-16}$ - το $e$ αυτό δεν έχει σχέση με το $e = 2.7182...$ που έχουμε ήδη συναντήσει, είναι απλά ένας συμβολισμός του "10 εις την"!), οπότε το αριθμητικό σφάλμα της ``roots`` είναι αμελητέο.

```python
# (a) Εύρεση των ριζών του πολυωνύμου z^3 + 8 - 8j = 0
coeffs1 = [1, 0, 0, 8-8*1j]
r1 = np.roots(coeffs1)
print(f"a. Οι ρίζες του πολυωνύμου είναι οι z1 = {r1}\n")

# Επιβεβαίωση
z1_true = [(8*np.sqrt(2))**(1/3)*np.exp(1j*(2*np.pi*k/3 + np.pi/4)) for k in range(0,3)]
print(f"a. Οι ρίζες του πολυωνύμου από τη θεωρία είναι οι z1_true = {np.asarray(z1_true)}\n")
```

```python
# (b) Εύρεση των ριζών του πολυωνύμου z^4 - 16 + 16j = 0
coeffs2 = [1, 0, 0, 0, -16+16*1j]
r2 = np.roots(coeffs2)
print(f"b. Οι ρίζες του πολυωνύμου είναι οι z2 = {r2}\n")

# Επιβεβαίωση
z2_true = [(16*np.sqrt(2))**(1/4)*np.exp(1j*((np.pi*k)/2 - np.pi/16)) for k in range(0,4)]
print(f"b. Οι ρίζες του πολυωνύμου από τη θεωρία είναι οι z2_true = {np.asarray(z2_true)}\n")
```

```python
# (c) Εύρεση των ριζών του πολυωνύμου z^5 + 32j = 0
coeffs3 = [1, 0, 0, 0, 0, 32*1j]
r3 = np.roots(coeffs3)
print(f"c. Οι ρίζες του πολυωνύμου είναι οι z3 = {r3}\n")

# Επιβεβαίωση
z3_true = [32**(1/5)*np.exp(1j*(2*np.pi*k/5 - np.pi/10)) for k in range(0,5)]
print(f"c. Οι ρίζες του πολυωνύμου από τη θεωρία είναι οι z3_true = {np.asarray(z3_true)}\n")
```
