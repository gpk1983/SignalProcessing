```python
import numpy as np                                          # Μαθηματική/αριθμητική βιβλιοθήκη
from IPython.display import Audio                           # Βιβλιοθήκη παιξίματος αρχείων ήχου
import pandas as pd                                         # Βιβλιοθήκη ανάγνωσης και επεξεργασίας αρχείων
```

## Άσκηση - Ένας Beethoven στο σπίτι σας!

Δημοφιλές πεδίο εφαρμογής της θεωρίας αποτελούν τα ηχητικά σήματα, οπότε ας παίξουμε με αυτά!

Αν θελήσουμε να δημιουργήσουμε στην `Python` ένα ημιτονοειδές σήμα συχνότητας $f_0=400$ Hz (νότα ΛΑ), διάρκειας $0.5$ sec (δηλ. $500$ ms), με μηδενική φάση μετατόπισης $\phi$ και μοναδιαίου πλάτους $A=1$, με συχνότητα δειγματοληψίας $f_s=16000$ Hz (που αντιστοιχεί σε καλής ποιότητας ηχογράφηση) θα γράψουμε τις ακόλουθες εντολές:

```python
fs = 16000    # συχνότητα δειγματοληψίας (fixed)
f0 = 400      # συχνότητα ημιτόνου
dur = 500     # διάρκεια σε ms
A = 1         # πλάτος ημιτόνου
t = np.arange(start=0, stop=dur/1000, step = 1/fs)   # άξονας χρόνου
signal = A*np.sin(2*np.pi*f0*t)                      # σήμα
```

Για να ακούσουμε αυτό που φτιάξαμε, πληκτρολογούμε

```python
Audio(signal, rate=fs)
```

Στην άσκηση αυτή θα συνθέσουμε τμήμα από το πολύ γνωστό κομμάτι του Beethoven, Fur Elise (https://www.youtube.com/watch?v=q9bU12gXUyM), από νότες πλήκτρων πιάνου.

(α') Με χρήση της βιβλιοθήκης `Pandas` μπορούμε να διαβάσουμε ένα CSV αρχείο `FurEliseData.csv` που μας δίνεται

```python
df = pd.read_csv('./files/FurEliseData.csv')
```

ως ένα `DataFrame`, και να το τυπώσουμε.

```python
print(df)
```

Όπως βλέπετε, υπάρχουν $4$ στήλες, `lefthand, righthand, ldur, rhdur` οι οποίες ανταποκρίνονται σε αριθμούς πλήκτρου πιάνου που παίζονται από το αριστερό χέρι (`lefthand`), με τις αντίστοιχες διάρκειές τους σε ms (`lhdur`), και σε αριθμούς πλήκτρου πιάνου που παίζονται με το δεξί χέρι (`righthand`), με τις αντίστοιχες επίσης διάρκειές τους σε ms (`rhdur`). Τα μηδενικά αντιστοιχούν σε παύσεις (δεν παίζουμε κάποιο πλήκτρο), αντίστοιχης διάρκειας.

Αυτούς μπορούμε να τους εισάγουμε σε ξεχωριστούς πίνακες της ``Python`` ως εξής:

```python
lefthand = df["lefthand"].values
righthand = df["righthand"].values
lhdur = df["lhdur"].values
rhdur = df["rhdur"].values

print(f"Τα πλήκτρα πιάνου του αριστερού χεριού είναι τα: \n {lefthand}")
print(f"Οι διάρκειες σε ms κάθε πλήκτρου του αριστερού χεριού είναι: \n {lhdur} ms")

print(f"Τα πλήκτρα πιάνου του δεξιού χεριού είναι τα: \n {righthand}")
print(f"Οι διάρκειες σε ms κάθε πλήκτρου του δεξιού χεριού είναι: \n {rhdur} ms")
```

(β') Μετατρέψτε από κάθε χέρι, κάθε αριθμό πλήκτρου πιάνου $n$ στην αντίστοιχη συνχότητα με τη σχέση
<br>
$$\Large f(n) = f_\mathrm{LA} \cdot 2^{\frac{n-49}{12}} = 440 \cdot 2^{\frac{n-49}{12}} $$
Στην `Python` αυτό γίνεται εύκολα ως

```python
# Μετατροπή πλήκτρου πιάνου σε νότα για το αριστερό χέρι
freq_lh = 440 * (2 ** ((lefthand-49)/12))

# Μετατροπή πλήκτρου πιάνου σε νότα για το αριστερό χέρι (συμπληρώστε)
freq_rh = 440 * (2 ** ((righthand-49)/12))
```

με αποτέλεσμα έναν πίνακα - διάνυσμα `freq` με τις αντίστοιχες συχνότητες. Οι παύσεις αναπαρίστανται στους πίνακες με τα πλήκτρα ως $(0)$. Θα μετατραπούν και αυτές - ενώ δεν πρέπει - όμως αυτό δε θα μας επηρεάσει, καθώς τις παύσεις θα τις χειριστούμε ξεχωριστά όπως θα δείτε παρακάτω.

(γ') Οι πίνακες `lhdur, rhdur` περιέχουν τη διάρκεια κάθε νότας σε ms για το αριστερό και το δεξί χέρι, αντίστοιχα. 

Ακολουθώντας τον παρακάτω κώδικα, δημιουργήστε τα αντίστοιχα ημίτονα με τη σωστή διάρκεια. Για τις παύσεις, θα δημιουργήσουμε ένα μηδενικό διάνυσμα με την ανάλογη διάρκεια.

```python
# Πόσα πλήκτρα έχουμε στο αριστερό χέρι?
N = len(lefthand)

# Συχνότητα δειγματοληψίας
fs = 16000

# Μια άδεια λίστα για να μαζεύουμε τα ημίτονα που θα φτιάχνουμε
signal_left = []

# Ξεκινάμε:
for i in range(1, N):
    
    # Ο άξονας χρόνου για κάθε πλήκτρο:
    t = np.arange(start=0, stop=lhdur[i]/1000, step=1/fs)
    
    if lefthand[i] == 0:
        # αν έχουμε παύση, δημιουργύμε ένα μηδενικό διάνυσμα διάρκειας lhdur[i] ms:
        signal_left = np.append(signal_left, np.zeros(len(t)))
    else:
        # φτιάξτε το αντίστοιχο ημίτονο με χρήση του διανύσματος freq_lh (χρησιμοποιήστε την np.sin() και το παράδειγμα στην αρχή της άσκησης)
        new_sin = np.sin(2*np.pi*freq_lh[i]*t)
        # "μαζέψτε" όλα τα ημίτονα στη λίστα signal_left
        signal_left = np.append(signal_left, new_sin)
```

(δ') Η μεταβλητή `signal_left` εδώ δηλώνεται ως μία λίστα (https://www.w3schools.com/python/python_lists.asp) της `Python`, στην οποία μπορούμε να *προσαρτήσουμε* (append) οποιαδήποτε μεταβλητή οποιουδήποτε μεγέθους. Αυτό είναι βολικό γιατί οι νότες δεν έχουν όλες την ίδια διάρκεια, και ένας πίνακας $N\times L$ θα ήταν δύσχρηστος μιας και το $L$ αλλάζει δυναμικά.

(ε') Επαναλάβετε τη διαδικασία για το δεξί χέρι, δηλ. τις μεταβλητές `righthand, rhdur` και αποθηκεύστε τα αποτελέσματά σας σε ξεχωριστές μεταβλητές. Παρατηρήστε ότι το αριστερό χέρι παίζει λιγότερες νότες από το δεξί.

```python
# Πόσα πλήκτρα έχουμε στο δεξί χέρι?
N = len(righthand)

# Συχνότητα δειγματοληψίας
fs = 16000

# Μια άδεια λίστα για να μαζεύουμε τα ημίτονα που θα φτιάχνουμε
signal_right = []

# Ξεκινάμε:
for i in range(1, N):
    
    # Ο άξονας χρόνου για κάθε πλήκτρο:
    t = np.arange(start=0, stop=rhdur[i]/1000, step=1/fs)
    
    if righthand[i] == 0:
        # αν έχουμε παύση, δημιουργύμε ένα μηδενικό διάνυσμα διάρκειας rhdur[i] ms:
        signal_right = np.append(signal_right, np.zeros(len(t)))
    else:
        # φτιάξτε το αντίστοιχο ημίτονο με χρήση του διανύσματος freq_rh  (χρησιμοποιήστε την np.sin() και το παράδειγμα στην αρχή της άσκησης)
        new_sin = np.sin(2*np.pi*freq_rh[i]*t)
        # "μαζέψτε" όλα τα ημίτονα στη λίστα signal_right
        signal_right = np.append(signal_right, new_sin)
```

(ζ) Δημιουργήστε το συνολικό σήμα, αθροίζοντας τα δύο σήματα ως

`sig = signal_left + signal_right`

```python
sig = signal_left + signal_right # Διαβάστε παρακάτω αν σας βγάλει error!
```

Αν τα έχετε κάνει όλα σωστά, η `Python` θα σας επιστρέψει σφάλμα, γιατί το ένα σήμα βγήκε λίγο μικρότερο από το άλλο, οπότε δε γίνεται να προστεθούν (να θυμάστε ότι οι πράξεις μεταξύ πινάκων πρέπει να τηρούν τους κανόνες διάστασης των πινάκων). Χρησιμοποιήστε την εντολή `len` και μετρήστε το μήκος του κάθε σήματος. Πόση είναι η διαφορά τους σε πλήθος τιμών;

```python
len_left = len(signal_left)
print(f"To μήκος σήματος αριστερού χεριού είναι {len_left} δείγματα.")

len_right = len(signal_right)
print(f"To μήκος σήματος δεξιού χεριού είναι {len_right} δείγματα.")

D = len_right - len_left
print(f"Το σήμα δεξιού χεριού είναι μεγαλύτερο κατά {D} δείγματα.")
```

(η') Για μια γρήγορη διόρθωση στο παραπάνω πρόβλημα, χρησιμοποιήστε τις εντολές

```python
D = len(signal_right) - len(signal_left)
signal_left = np.append(signal_left, np.zeros(D))
sig = signal_left + signal_right
```

όπου απλά συμπληρώνουμε με μηδενικά το μικρότερο διάνυσμα για να γίνει ίσου "μεγέθους" με το μεγαλύτερο.

(θ') Μπορείτε τώρα να ακούσετε το σήμα σας!

```python
sig = signal_left + signal_right

# Ακούμε (χαμηλώστε τον ήχο)
Audio(sig, rate=fs)
```

Να σημειωθεί ότι ένα **πραγματικό πλήκτρο πιάνου** δεν παράγει *ένα* ημίτονο, όπως κάναμε εμείς, αλλά ένα **άθροισμα από ημίτονα**, με συχνότητες σχεδόν ακέραιες πολλαπλάσιες της θεμελιώδους συχνότητας και διαφορετικά μεταξύ τους πλάτη. Με άλλα λόγια, παράγει - προσεγγιστικά - **μια σειρά Fourier**, με πεπερασμένο πλήθος όρων, την οποία θα μελετήσουμε σύντομα. Επίσης, σε ένα πραγματικό πιάνο, υπάρχουν πάρα πολλές λεπτομέρειες που μας χαρίζουν τον πλούσιο ήχο του, οι οποίες δεν είναι απλό να μοντελοποιηθούν.
