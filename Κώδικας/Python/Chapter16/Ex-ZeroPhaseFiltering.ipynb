## Άσκηση - Φιλτράρισμα Μηδενικής Φάσης
(α') Γνωρίζετε από τη θεωρία σας ότι όταν περνάμε ένα σήμα μέσα από ένα ΓΧΑ σύστημα, τότε το μεταβάλλουμε κατά πλάτος και κατά φάση. Ο σχεδιασμός ενός συστήματος (φίλτρου, στο εξής) εστιάζεται κυρίως στην αλλαγή του φάσματος πλάτους του σήματος εισόδου, χωρίς να δίνεται ιδιαίτερη βαρύτητα στην αλλαγή της φάσης του. 

Παρ'όλα αυτά, υπάρχουν δυο - τουλάχιστον - στρατηγικές για να αποφύγουμε την αλλαγή φάσης του σήματος εισόδου, ακόμα κι όταν το φίλτρο μας δεν είναι μηδενικής ή σταθερής φάσης.

Στην προσπάθειά σας, **συνίσταται θερμά να μετατρέψετε τις εξισώσεις στο χρόνο στο χώρο του μετασχ. Fourier**.

Συγκεκριμένα, η 1η στρατηγική αναφέρει ότι περνάμε το σήμα μας μέσα από το επιθυμητό φίλτρο, παίρνοντας μια έξοδο $x_1[n]$. Στη συνέχεια, η είσοδος αναστρέφεται χρονικά, $x[-n]$, και ξαναπερνά από το επιθυμητό φίλτρο, λαμβάνοντας την έξοδο $x_2[n]$. Η τελική έξοδος δίνεται από τη σχέση

$$y[n] = \frac{1}{2}(x_1[n] + x_2[-n])$$

Με αυτόν τον τρόπο, παρ'όλο που το φίλτρο $h[n]$ μπορεί να είναι ΜΗ σταθερής φάσης, η φάση του δεν επηρεάζει την έξοδο $y[n]$. Η παραπάνω διαδικασία μπορεί να απλοποιηθεί χρησιμοποιώντας ένα ισοδύναμο φίλτρο, το $h_1[n]$, που εκτελεί τις ίδιες διαδικασίες με το διπλό "πέρασμα" που μόλις είδαμε. Ζητείται αυτή ακριβώς η συνολική κρουστική απόκριση $h_1[n]$.

Αντίστοιχα, η 2η στρατηγική αναφέρει ότι περνάμε το σήμα μας μέσα από το επιθυμητό φίλτρο, παίρνοντας μια έξοδο $x_1[n]$. Στη συνέχεια, η **έξοδος** αυτή αναστρέφεται χρονικά, $x_1[-n]$, και ξαναπερνά από το επιθυμητό φίλτρο, λαμβάνοντας την έξοδο $x_2[n]$. Η τελική έξοδος δίνεται από τη σχέση

$$y[n] = x_2[-n]$$

Με αυτόν τον τρόπο, παρ'όλο που το φίλτρο $h[n]$ μπορεί να είναι ΜΗ σταθερής φάσης, η φάση του δεν επηρεάζει την έξοδο $y[n]$. Η παραπάνω διαδικασία μπορεί να απλοποιηθεί χρησιμοποιώντας ένα ισοδύναμο φίλτρο, το $h_2[n]$, που εκτελεί τις ίδιες διαδικασίες με το διπλό "πέρασμα" που μόλις είδαμε. Ζητείται αυτή ακριβώς η συνολική κρουστική απόκριση $h_2[n]$.


<u>**Απάντηση**</u>


### i) Στρατηγική I<u></u>
Χρησιμοποιώντας τις ιδιότητες της γραμμικότητας και της αντιστροφής στο χρόνο στον μετασχηματισμό Fourier, προκύπτει:

$$
\left.
    \begin{array}{lr}
        x_{1}[n] = x[n] * h[n]\\
        x_{2}[n] = x[-n] * h[n]\\
        y[n] = h_{1}[n] * x[n]\\
        y[n] = \frac{x_{1}[n] + x_{2}[-n]}{2}
    \end{array}
\right \}
\overset{F}{\Longrightarrow}

\begin{array}{lr}
        X_{1}(e^{j\omega}) = X(e^{j\omega})H(e^{j\omega})\\
        X_{2}(e^{j\omega}) = X(e^{-j\omega})H(e^{j\omega})\\
        Y(e^{j\omega}) = H_{1}(e^{j\omega})X(e^{j\omega})\\
        Y(e^{j\omega}) = \frac{X_{1}(e^{j\omega}) + X_{2}(e^{-j\omega})}{2}
\end{array}
$$
Άρα
$$
\begin{align}
H_{1}(e^{j\omega})X(e^{j\omega}) &= \frac{X_{1}(e^{j\omega})+ X_{2}(e^{-j\omega})}{2} \\
H_{1}(e^{j\omega})X(e^{j\omega}) &= \frac{X(e^{j\omega})H(e^{j\omega})+ X(e^{j\omega})H(e^{-j\omega})}{2}\\
H_{1}(e^{j\omega}) &= \frac{H(e^{j\omega})+ H(e^{-j\omega})}{2} \overset{F^{-1}}{\Longrightarrow} h_{1}[n] = \frac{h[n]+h[-n]}{2}\\
\end{align}
$$
Αν το $h[n]$ είναι πραγματικό, ισχύει επιπλεόν:
$$H_{1}(e^{j\omega}) = \frac{H(e^{j\omega}) + H(e^{-j\omega})}{2} = \frac{H(e^{j\omega}) + H^{*}(e^{j\omega})}{2} = \frac{2\Re\{ H(e^{j\omega})\}}{2} = \Re \{H(e^{j\omega})\} = |H(e^{j\omega})| \cos (\phi(\omega))$$
σύμφωνα με τις ιδιότητες των μιγαδικών αριθμών. Έτσι: $\measuredangle H_{1} = \measuredangle |H(e^{j\omega})| \cos (\phi(\omega)) = 0$ ή $\pm \pi$, ανάλογα με την τιμή του $\phi(\omega)$. Άρα, έχει μηδενική ή σταθερή φάση.


### ii) Στρατηγική II<u></u>
Χρησιμοποιώντας τις ιδιότητες της γραμμικότητας και της αντιστροφής στο χρόνο στον μετασχηματισμό Fourier, προκύπτει:
$$
\left.
    \begin{array}{lr}
        x_{1}[n] = x[n] * h[n]\\
        x_{2}[n] = x[-n] * h[n]\\
        y[n] = h_{2}[n] * x[n]\\
        y[n] = x_{2}[-n]
    \end{array}
\right \}
\overset{F}{\Longrightarrow}
\begin{array}{lr}
        X_{1}(e^{j\omega}) = X(e^{j\omega})H(e^{j\omega})\\
        X_{2}(e^{j\omega}) = X_{1}(e^{-j\omega})H(e^{j\omega})\\
        Y(e^{j\omega}) = H_{2}(e^{j\omega})X(e^{j\omega})\\
        Y(e^{j\omega}) = X_{2}(e^{-j\omega})
\end{array}
$$
Άρα
$$
\begin{align}
H_{2}(e^{j\omega})X(e^{j\omega}) &= X_{2}(e^{-j\omega})\\
H_{2}(e^{j\omega})X(e^{j\omega}) &= X_{1}(e^{j\omega})H(e^{-j\omega})\\
H_{2}(e^{j\omega})X(e^{j\omega}) &= X(e^{j\omega})H(e^{j\omega})H(e^{-j\omega})\\
H_{2}(e^{j\omega}) &= H(e^{j\omega})H(e^{-j\omega}) \overset{F^{-1}}{\Longrightarrow} h_{2}[n] = h[n] * h[-n]
\end{align}
$$
Αν το $h[n]$ είναι πραγματικό, ισχύει επιπλέον:
$$H_{2}(e^{j\omega}) = H(e^{j\omega})H(e^{-j\omega}) = H(e^{j\omega})H^{*}(e^{j\omega}) = |H(e^{j\omega})|^{2}$$
σύμφωνα με τις ιδιότητες των μιγαδικών αριθμών. Έτσι: $\measuredangle H_{2} = \measuredangle |H(e^{j\omega})|^{2} = 0$, άρα έχει μηδενική φάση.

(β') Ας εφαρμόσουμε τώρα τα παραπάνω σε ένα πραγματικό πρόβλημα στη Python. Με την εντολή ``np.load`` θα διαβάσετε το αρχείο ``FIRfilter.mat`` που σας δίνεται.

Τώρα έχετε ένα μικρό δείγμα από ένα πραγματικό σήμα φωνής στη μεταβλητή ``fr1``, και τους συντελεστές ενός χαμηλοπερατού φίλτρου πεπερασμένης κρουστικής απόκρισης (FIR) $71$ συντελεστών στη μεταβλητή $B$. 

Θέλουμε να αφαιρέσουμε τις υψηλές συχνότητες από το σήμα της φωνής χωρίς αυτό να μετακινηθεί στο χρόνο (χωρίς δηλαδή, να υποστεί μεταβολή φάσης λόγω του φίλτρου). Αυτή η απαίτηση ονομάζεται στην επεξεργασία σήματος ως **φιλτράρισμα μηδενικής φάσης - zero-phase filtering**. 

Ακολουθήστε τα παρακάτω βήματα και δουλέψτε στο πεδίο του **χρόνου**. Αν τα κάνετε όλα όπως πρέπει θα πάρετε στο τέλος ένα γράφημα όπως στο παρακάτω σχήμα.

![outputFIR.png](./files/outputFIR.png)

Παρακάτω παρουσιάζουμε τα γραφήματα που παρουσιάζουν τα $3$ επιμέρους βήματα ξεχωριστά, δηλαδή ξεχωριστά το αποτέλεσμα του κάθε φιλτραρίσματος:

```python
m = io.loadmat('./files/FIRfilter.mat')

# Φορτώνουμε τις μεταβλητές
fr1 = m['fr1'][0]
B = m['B'][0]

# Τυπώνουμε το αρχικό σήμα
fig1 = plt.figure(figsize=(12,4))
plt.title('Signal fr1')
plt.plot(fr1)
plt.grid(True)
plt.xlabel('Samples')
plt.show()
```

i. Φιλτράρετε απ΄ ευθείας το σήµα µε την εντολή ``sig.lfilter``, όπως γνωρίζετε από προηγούμενα ερωτήματα.

```python
# Φιλτράρισμα χωρίς καμία στρατηγική
y0 = sig.lfilter(B, 1, fr1)

plt.figure(figsize=(12,4))
plt.plot(fr1)
plt.plot(y0, linewidth=2)
plt.grid(True)
plt.xlabel('Δείγματα')
plt.title('Φιλτράρισμα του αρχικού σήματος χωρίς χρήση κάποιας στρατηγικής')
plt.legend(['Αρχικό Σήμα','Φιλτραρισμένο χωρίς στρατηγικές'], loc='upper left')
plt.xlim(0, 400)
plt.ylim(-1500, 1500)
plt.show()
```

ii. Φιλτράρετε το σήμα ακολουθώντας τη στρατηγική Ι.

```python
# Φιλτράρισμα με τη 1η στρατηγική
x2 = np.fliplr(fr1.reshape(1, len(fr1)))
x1n = sig.lfilter(B, 1, fr1)
x2n = sig.lfilter(B, 1, x2)

y1 = 1/2 * (x1n + np.fliplr(x2n))[0]

plt.figure(figsize=(12,4))
plt.plot(fr1)
plt.plot(y1, color=[0.9290, 0.6940, 0.1250], linewidth=2)
plt.grid(True)
plt.xlabel('Δείγματα')
plt.title('Φιλτράρισμα του σήματος χρησιμοποιώντας τη 1η στρατηγική')
plt.legend(['Αρχικό Σήμα','Φιλτραρισμένο με την 1η Στρατηγική'], loc= 'upper right')
plt.show()
```

iii. Φιλτράρετε το σήμα ακολουθώντας τη στρατηγική ΙΙ.

```python
# Φιλτράρισμα του αρχικού σήματος χρησιμοποιώντας τη 2η στρατηγική
x2 = np.fliplr(fr1.reshape(1, len(fr1)))[0]
x1n = sig.lfilter(B, 1, fr1)
x2n = sig.lfilter(B, 1, np.fliplr(x1n.reshape(1, len(x1n))))
x2n = x2n[0]
y2 = np.fliplr(x2n.reshape(len(x2n), 1).T)[0]

plt.figure(figsize=(12,4))
plt.plot(fr1)
plt.plot(y2, color=[0.4940, 0.1840, 0.5560], linewidth=2)
plt.grid(True)
plt.xlabel('Δείγματα')
plt.title('Φιλτράρισμα του σήματος χρησιμοποιώντας τη 2η στρατηγική')
plt.legend(['Αρχικό Σήμα','Φιλτραρισμένο με την 2η Στρατηγική'], loc= 'upper left')
plt.show()
```

iv. Ας παρουσιάσουμε όλα τα αποτελέσματα μαζί σε ένα διάγραμμα. 

```python
# Σύγκριση των αποτελεσμάτων των τριών στρατηγικών
plt.figure(figsize=(12,4))
plt.plot(fr1)
plt.plot(y0, linewidth=2)
plt.plot(y1, linewidth=2)
plt.plot(y2, linewidth=2)
plt.grid(True)
plt.xlabel('Δείγματα')
plt.title('Φιλτράρισμα του σήματος με διάφορες μεθόδους')
plt.legend(['Αρχικό Σήμα','Φιλτραρισμένο χωρίς στρατηγικές','Φιλτραρισμένο με την 1η Στρατηγική','Φιλτραρισμένο με την 2η Στρατηγική'], loc='upper left')
plt.show()
```

Η στρατηγική Ι και το φιλτράρισμα χωρίς στρατηγική φαίνεται ότι έχουν παρόμοιο αποτέλεσμα μεταξύ τους, όμως έχουν αρκετές διαφορές σε σχέση με το αρχικό σήμα, λόγω διαφορετικής φάσης.
